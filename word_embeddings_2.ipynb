```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Embedding,Flatten
from tensorflow.keras.datasets import imdb
```

```python
vocab_size = 10000
max_len = 200
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=vocab_size)
```

```python
print(X_train[0])
```

```python
padded_train_review = pad_sequences(X_train,maxlen=max_len,padding="post")
padded_test_review = pad_sequences(X_test,maxlen=max_len,padding="post")
```

```python
padded_train_review[:5]
```

```python
y_train[:5]
```

```python
embedded_vec_size = 16

model = Sequential([
    Embedding(vocab_size,embedded_vec_size,input_length=max_len,name="embedding"),
    Flatten(),
    Dense(1,activation="sigmoid")
])
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=['accuracy']
)
model.fit(padded_train_review,y_train,epochs=30)
```

```python
model.get_layer("embedding").get_weights()[0]
```

```python
model.evaluate(padded_test_review,y_test)
```

```python
model.evaluate(padded_train_review,y_train)
```
